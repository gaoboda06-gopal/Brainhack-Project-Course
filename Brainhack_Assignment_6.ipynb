```python
%%writefile useful_functions.py

UNICODE_SIZE = 1114112

def encrypt_letter(letter, key):
    encrypted_index = (ord(letter) + ord(key)) % UNICODE_SIZE
    return chr(encrypted_index)

def decrypt_letter(letter, key):
    decrypted_index = (ord(letter) - ord(key)) % UNICODE_SIZE
    return chr(decrypted_index)

def process_message(message, key, encrypt):
    result = []
    key_length = len(key)
    for i, letter in enumerate(message):
        key_letter = key[i % key_length]
        if encrypt:
            result.append(encrypt_letter(letter, key_letter))
        else:
            result.append(decrypt_letter(letter, key_letter))
    return ''.join(result)

if __name__ == "__main__":
    message = "word"
    key = "key"
    encrypted_msg = process_message(message, key, encrypt=True)
    decrypted_msg = process_message(encrypted_msg, key, encrypt=False)
    print(f"Original  : {message}")
    print(f"Encrypted : {encrypted_msg}")
    print(f"Decrypted : {decrypted_msg}")
    if message == decrypted_msg:
        print("Test passed")
    else:
        print("Test failed")
```

```python
%%writefile cypher_script.py

import argparse
from useful_functions import process_message

def parse_arguments():
    parser = argparse.ArgumentParser(
        description="Vigenere cipher: encrypt or decrypt a text file."
    )
    parser.add_argument("-i", required=True, help="Path to the input text file")
    parser.add_argument("-o", required=True, help="Path to the output text file")
    parser.add_argument("-k", required=True, help="Cipher key string")
    parser.add_argument("-m", required=True, choices=["encryption", "decryption"],
                        help='Mode: "encryption" or "decryption"')
    return parser.parse_args()

def main():
    args = parse_arguments()
    with open(args.i, "r", encoding="utf-8") as f:
        message = f.read()
    encrypt = (args.m == "encryption")
    result  = process_message(message, args.k, encrypt)
    with open(args.o, "w", encoding="utf-8") as f:
        f.write(result)
    print(f"Done! '{args.m}' complete.")
    print(f"  Input  : {args.i}")
    print(f"  Output : {args.o}")
    print(f"  Key    : {args.k}")

if __name__ == "__main__":
    main()
```
